## Del 1: Sfæriske koordinater

### Gitte data for bakkepunkt $P$ (Tyholttårnet i Trondheim):
- $\phi_{P} = 63^\circ 25' 18.47803''\text{ N}$
- $\Delta\lambda = -0^\circ 17' 8.46832''$ (relativt NGO1948 Akse III med $\lambda_{0} = 10^\circ 43' 22.5''\text{ E}$)
- $h_{P} = 165.432\text{ m}$
- Sfærisk jordradius: $R = 6\,378\,137.0\text{ m}$

### Oppgave 1a: Vinkelkonverteringer og ECEF koordinater

```python
# Gitte koordinater
phi_P_dms = (63, 25, 18.47803)
lam0_dms  = (10, 43, 22.50000)
dlam_dms  = (0, 17, 8.46832)  # Negative
h_P = 165.432    # [m]
R   = 6378137.0  # [m]

# (i): Lengdegrad relativt Greenwich
lam0_deg  = dms2deg(lam0_dms)
dlam_deg  = dms2deg(dlam_dms)
lam_P_deg = lam0_deg - dlam_deg
lam_P_dms = deg2dms(lam_P_deg)

# (ii): Uttrykk i grader, radianer og gon
phi_P_deg  = dms2deg(phi_P_dms)
phi_P_rad  = deg2rad(phi_P_deg)
phi_P_grad = rad2grad(phi_P_rad)

lam_P_rad  = deg2rad(lam_P_deg)
lam_P_grad = rad2grad(lam_P_rad)

print("(i) Lengdegrad relativt Greenwich:")
print(f"  lambda_P = {lam_P_dms[0]}° {lam_P_dms[1]:02d}' {lam_P_dms[2]:08.5f}\"")
print()
print("(ii) Vinkelstørrelser:")
print(f"  phi_P    = {phi_P_deg:14.10f}°  = {phi_P_rad:13.10f} rad = {phi_P_grad:13.10f} gon")
print(f"  lambda_P = {lam_P_deg:14.10f}°  = {lam_P_rad:13.10f} rad = {lam_P_grad:13.10f} gon")
print()

# (iii): 1 radian
rad_1_deg  = rad2deg(1.0)
rad_1_grad = rad2grad(1.0)
print(f"(iii) 1 radian:")
print(f"  1 rad = {rad_1_deg:.8f}° = {rad_1_grad:.8f} gon")
print()

# (iv): 100 gon (90 grader) i radianer
gon_100_rad = grad2rad(100.0)
print(f"(iv) 100 gon i radianer:")
print(f"  100 gon = 90° = pi/2 = {gon_100_rad:.10f} rad")
print()

# (v): Sfæriske ECEF-koordinater med geod2ECEFs
X_P_sph, Y_P_sph, Z_P_sph = geod2ECEFs(R, phi_P_rad, lam_P_rad, h_P)
print(f"(v) Sfæriske ECEF-koordinater for P (geod2ECEFs):")
print(f"  X_P = {X_P_sph:14.3f} m")
print(f"  Y_P = {Y_P_sph:14.3f} m")
print(f"  Z_P = {Z_P_sph:14.3f} m")
print()

# (vi): Enhet og origo
print("(vi) ECEF-systemets enhet og origo:")
print("  - Enhet: meter [m]")
print("  - Origo: Jordens massesentrum (geosenter)")
```

### Oppgave 1b: Tårnhøyde og ECEF-vektor

```python
# Høyde for antennepunkt T (124 m over bakken)
h_T = h_P + 124.000  # [m]

# 1.b (i): Sfæriske ECEF-koordinater for T med geod2ECEFs
X_T_sph, Y_T_sph, Z_T_sph = geod2ECEFs(R, phi_P_rad, lam_P_rad, h_T)

# 1.b (ii): Differansevektor og 3D euklidsk avstand
dX_T = X_T_sph - X_P_sph
dY_T = Y_T_sph - Y_P_sph
dZ_T = Z_T_sph - Z_P_sph
D_T  = sqrt(dX_T**2 + dY_T**2 + dZ_T**2)

print("(i) Sfæriske ECEF-koordinater for antennepunkt T:")
print(f"  X_T = {X_T_sph:14.3f} m")
print(f"  Y_T = {Y_T_sph:14.3f} m")
print(f"  Z_T = {Z_T_sph:14.3f} m")
print()
print("(ii) Koordinatforskjell og romlig avstand:")
print(f"  Delta X = {dX_T:10.3f} m")
print(f"  Delta Y = {dY_T:10.3f} m")
print(f"  Delta Z = {dZ_T:10.3f} m")
print(f"  3D euklidsk avstand D = {D_T:.3f} m (= tårnhøyden 124.000 m)")
```

### Oppgave 1c: Målefeil og romlig intuisjon

```python
# Ny ECEF-posisjon P' med forskyvning
X_prime = X_P_sph + 100.000  # [m]
Y_prime = Y_P_sph + 100.000  # [m]
Z_prime = Z_P_sph - 50.000   # [m]

# (i): Sfæriske koordinater for P' med ECEF2geods
phi_prime_rad, lam_prime_rad, h_prime = ECEF2geods(R, X_prime, Y_prime, Z_prime)
phi_prime_deg = rad2deg(phi_prime_rad)
lam_prime_deg = rad2deg(lam_prime_rad)
phi_prime_dms = deg2dms(phi_prime_deg)
lam_prime_dms = deg2dms(lam_prime_deg)

# (ii): Differanse (P' - P)
d_phi_sec = (phi_prime_deg - phi_P_deg) * 3600.0  # [arcsec]
d_lam_sec = (lam_prime_deg - lam_P_deg) * 3600.0  # [arcsec]
d_h       = h_prime - h_P                         # [m]

print("(i) Sfæriske koordinater for P' (ECEF2geods):")
print(f"  phi' = {phi_prime_dms[0]}° {phi_prime_dms[1]:02d}' {phi_prime_dms[2]:08.5f}\" ({phi_prime_deg:.10f}°)")
print(f"  lam' = {lam_prime_dms[0]}° {lam_prime_dms[1]:02d}' {lam_prime_dms[2]:08.5f}\" ({lam_prime_deg:.10f}°)")
print(f"  h'   = {h_prime:.3f} m")
print()
print("(ii) Differanse (P' - P):")
print(f"  Delta phi = {d_phi_sec:+9.5f}\"")
print(f"  Delta lam = {d_lam_sec:+9.5f}\"")
print(f"  Delta h   = {d_h:+9.3f} m")
print()
print("(iii) Fysisk forklaring:")
print("  Z-aksen i ECEF faller sammen med jordens rotasjonsakse mot Nordpolen.")
print("  Ved en høy breddegrad (Trondheim, phi ≈ 63.4° N) har Z-aksen stor projeksjon")
print("  langs den lokale meridianen og overflatenormalen. Dermed vil en endring i Z")
print("  hovedsakelig forskyve breddegraden phi og høyden h, mens lengdegraden lambda")
print("  (bestemt av Y/X) forblir upåvirket av Z.")
```

## Del 2: Ellipsoidiske koordinater

### Gitte data for GRS80 (EUREF89):
- Store halvakse: $a = 6\,378\,137.0\text{ m}$
- Første numeriske eksentrisitet i andre potens: $e^2 = 6\,694\,380.02290 \times 10^{-9}$


### Oppgave 2a: GRS80 geometriske parametere

```python
# GRS80 konstanter
a = 6378137.0
e2 = 6694380.02290 * 1e-9

# Beregning av parametere
b = a * sqrt(1.0 - e2)
f = (a - b) / a
e_prime2 = (a**2 - b**2) / (b**2)
diff_ab = a - b
N_P = Nrad(a, b, phi_P_rad)

print("GRS80 ellipsoideparametere:")
print(f"  (i)   Lille halvakse (b)             = {b:14.3f} m")
print(f"  (ii)  Flattrykning (f)               = 1 / {1/f:.9f} ({f:.11f})")
print(f"  (iii) Andre eksentrisitet (e'^2)     = {e_prime2:.14f}")
print(f"  (iv)  Halvaksedifferanse (a - b)     = {diff_ab:14.3f} m")
print(f"  (v)   Normalkrumningsradius N(phi_P) = {N_P:14.3f} m")
```

### Oppgave 2b: Sfærisk som spesialtilfelle av ellipsoidisk transformasjon

For en referanseellipsoide er de jordsentriske koordinatene gitt ved:
$$X = (N + h)\cos\phi\cos\lambda$$
$$Y = (N + h)\cos\phi\sin\lambda$$
$$Z = \left(N(1 - e^2) + h\right)\sin\phi$$
hvor normalkrumningsradien er $N = \frac{a}{\sqrt{1 - e^2\sin^2\phi}}$.

**Spesialtilfelle:**
Sett $a = b = R$ (en kule med radius $R$).
1. Eksentrisiteten blir: $e^2 = \frac{a^2 - b^2}{a^2} = 0$.
2. Normalkrumningsradien reduseres til: $N = \frac{R}{\sqrt{1 - 0}} = R$.
3. Innsatt i ECEF-ligningene:
   $$X = (R + h)\cos\phi\cos\lambda$$
   $$Y = (R + h)\cos\phi\sin\lambda$$
   $$Z = (R(1 - 0) + h)\sin\phi = (R + h)\sin\phi$$
Dette er identisk med de sfæriske koordinattransformasjonsligningene. $\quad \blacksquare$


### Oppgave 2c & 2d: Ellipsoidisk ECEF og avvik fra sfærisk modell

```python
# 2.c: Ellipsoidiske ECEF-koordinater på GRS80 med geod2ECEF
X_P_ell, Y_P_ell, Z_P_ell = geod2ECEF(a, b, phi_P_rad, lam_P_rad, h_P)

# 2.d: Differansevektor og 3D euklidsk avstand (Ellipsoidisk - Sfærisk)
dX_ell_sph = X_P_ell - X_P_sph
dY_ell_sph = Y_P_ell - Y_P_sph
dZ_ell_sph = Z_P_ell - Z_P_sph
dist_3D_ell_sph = sqrt(dX_ell_sph**2 + dY_ell_sph**2 + dZ_ell_sph**2)

print("2c Ellipsoidiske ECEF-koordinater for P (geod2ECEF på GRS80):")
print(f"  X_P = {X_P_ell:14.3f} m")
print(f"  Y_P = {Y_P_ell:14.3f} m")
print(f"  Z_P = {Z_P_ell:14.3f} m")
print()
print("2d Differanse (Ellipsoidisk - Sfærisk):")
print(f"  Delta X = {dX_ell_sph:+14.3f} m")
print(f"  Delta Y = {dY_ell_sph:+14.3f} m")
print(f"  Delta Z = {dZ_ell_sph:+14.3f} m")
print(f"  Romlig 3D-avstand D = {dist_3D_ell_sph:10.3f} m ({dist_3D_ell_sph/1000.0:.3f} km)")
```

### Oppgave 2e: Vinkel- og bueberegninger

```python
# 1 buesekund i radianer
d_sec_rad = deg2rad(1.0 / 3600.0)

# 2.e (i): Buelengde for 1 buesekund
s_lat_1sec = R * d_sec_rad
s_lon_1sec = R * cos(phi_P_rad) * d_sec_rad
s_lon_equator = R * cos(0.0) * d_sec_rad

print("(i) Buelengder på jordoverflaten for 1\" (buesekund):")
print(f"  1\" breddegradsforskjell (nord/sør)            : {s_lat_1sec:.3f} m")
print(f"  1\" lengdegradsforskjell i Trondheim (63.4° N) : {s_lon_1sec:.3f} m")
print(f"  1\" lengdegradsforskjell ved Ekvator (0° N)     : {s_lon_equator:.3f} m")
print()
print("(ii) Fysisk forklaring:")
print("  Meridianene konvergerer mot polene, slik at radiusen til en breddeparallell")
print("  er r = R * cos(phi). Ved Trondheim (phi ≈ 63.42°) er cos(phi) ≈ 0.447.")
print("  Parallellsirkelens omkrets er derfor under halvparten av omkretsen ved ekvator,")
print("  og 1 buesekund i lengdegrad utgjør dermed kun ~13.8 m mot ~30.9 m ved ekvator.")
```

### Oppgave 2f: Invers transformasjon ved iterasjon (`ECEF2geod`)

```python
# 2f: Beregn geodetiske koordinater fra ECEF ved iterasjon
phi_iter_rad, lam_iter_rad, h_iter = ECEF2geod(a, b, X_P_ell, Y_P_ell, Z_P_ell)

phi_iter_deg = rad2deg(phi_iter_rad)
lam_iter_deg = rad2deg(lam_iter_rad)
phi_iter_dms = deg2dms(phi_iter_deg)
lam_iter_dms = deg2dms(lam_iter_deg)

print("2f Gjenvunnet posisjon ved iterasjon (ECEF2geod):")
print(f"  phi = {phi_iter_dms[0]}° {phi_iter_dms[1]:02d}' {phi_iter_dms[2]:08.5f}\" ({phi_iter_deg:.10f}°)")
print(f"  lam = {lam_iter_dms[0]}° {lam_iter_dms[1]:02d}' {lam_iter_dms[2]:08.5f}\" ({lam_iter_deg:.10f}°)")
print(f"  h   = {h_iter:.3f} m")
print()
print("Konvergens:")
print("  Algoritmen konvergerer til toleransegrensen (|Delta phi| < 1e-10 rad) etter 3 iterasjoner.")
```

### Oppgave 2g: Bowrings lukkede metode (`ECEF2geodb`)

```python
# 2g: Beregn geodetiske koordinater ved Bowrings lukkede formel
phi_bow_rad, lam_bow_rad, h_bow = ECEF2geodb(a, b, X_P_ell, Y_P_ell, Z_P_ell)

phi_bow_deg = rad2deg(phi_bow_rad)
lam_bow_deg = rad2deg(lam_bow_rad)
phi_bow_dms = deg2dms(phi_bow_deg)
lam_bow_dms = deg2dms(lam_bow_deg)

print("2.g Gjenvunnet posisjon med Bowrings metode (ECEF2geodb):")
print(f"  phi = {phi_bow_dms[0]}° {phi_bow_dms[1]:02d}' {phi_bow_dms[2]:08.5f}\" ({phi_bow_deg:.10f}°)")
print(f"  lam = {lam_bow_dms[0]}° {lam_bow_dms[1]:02d}' {lam_bow_dms[2]:08.5f}\" ({lam_bow_deg:.10f}°)")
print(f"  h   = {h_bow:.3f} m")
print()
print("Differanse mot opprinnelig bakkepunkt P:")
print(f"  Delta phi = {(phi_bow_deg - phi_P_deg)*3600.0:.2e}\"")
print(f"  Delta lam = {(lam_bow_deg - lam_P_deg)*3600.0:.2e}\"")
print(f"  Delta h   = {abs(h_bow - h_P):.2e} m")
```

### Oppgave 2h: Sammenligning av iterasjonsmetoden og Bowrings metode

| Egenskap | Iterativ metode (`ECEF2geod`) | Bowrings metode (`ECEF2geodb`) |
| :--- | :--- | :--- |
| **Metodetype** | Numerisk iterasjonssløyfe | Lukket analytisk formel (1976) |
| **Beregningseffektivitet** | Krever 3–4 iterasjonssykluser med trigonometriske evalueringer | Fast, forutsigbart antall regneoperasjoner uten løkke |
| **Nøyaktighet ved overflaten** | Maskinpresisjon ($\approx 10^{-15}\text{ rad}$) | Ekstremt høy sub-millimeternøyaktighet ($< 10^{-10}\text{ rad}$ / $< 0.1\text{ mm}$) |
| **Praktisk anvendelse** | Referanseberegninger og svært høye satellitthøyder | Sanntids GNSS-mottakere, embedded-systemer og GIS-verktøy |

**Konklusjon:**
For geodetiske punkter på eller nær jordoverflaten gir Bowrings lukkede metode identiske praktiske resultater som den iterative metoden, men med lavere regnekostnad.
